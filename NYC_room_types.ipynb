## **Data Cleaning & Feature Engineering**

- Drop columns that are pure identifiers or free text and carry no generalized meaning for a model (`id`, `name`, `host_id`, `host_name`, `last_review`)

- Fill missing (NaN) `reviews_per_month` with 0

- Cap extreme outliers in `price` and `minimum_nights` using percentile clipping, so that handful of data-entry errors (e.g., $10K/night or 1250 minimum nights) don't distort the model

- Seperate features (X) from target (y)

```python
df_clean = df.drop(columns = ['id', 'name', 'host_id', 'host_name', 'last_review'])
```

```python
df_clean
```

```python
df_clean['reviews_per_month'] = df_clean['reviews_per_month'].fillna(0)
```

```python

df_clean
```

```python
price_cap = df_clean['price'].quantile(0.99)
nights_cap = df_clean['minimum_nights'].quantile(0.99)


df_clean['price'] = df_clean['price'].clip(upper=price_cap)
df_clean['minimum_nights'] = df_clean['minimum_nights'].clip(upper=nights_cap)

"""
df_clean = df_clean[df_clean['price'] < price_cap]
df_clean = df_clean[df_clean['minimum_nights'] < nights_cap]
"""
```

```python
df_clean
```

```python
X = df_clean.drop(columns = ['room_type'])
y = df_clean['room_type']
```

## Train / Test Split
- 20% of data -> test set, never touched during model selection or tuning, only used once, at very end, to report final performance
- `stratify=y` keeps same class proportions in both splits, which matters because target is imbalanced.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.33, random_state = 42, stratify = y)
```

```python
# Stratify : ensures whenever we split or sample our data, every class (group) original proportion is maintained
```

## Preprocessing : `ColumnTransformer` + `Pipeline`

Production ML code should never manually transform train/test data with seperate lines, it is error prone and leaks information.
Instead we build a single, reusable `ColumnTransformer`:

- Numeric features -> median imputation + standard scaling
- Categorical features -> most-frequent imputation + one-hot encoding

This transformer will be the first step of every model pipeline below, so preprocessing is learned only on training data and applied consistently everywhere (no data leakage).

```python
df_clean[numerical_cols].skew() #skewness
```

```python
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder, PowerTransformer
from sklearn.compose import ColumnTransformer
```

```python
numerical_cols = ['latitude', 'longitude', 'price', 'minimum_nights', 'number_of_reviews',
                  'reviews_per_month', 'calculated_host_listings_count', 'availability_365']

categorical_cols = ['neighbourhood_group', 'neighbourhood']

# 1. pipeline -> for numeric columns

numeric_pipeline = Pipeline(steps=[
    ('impute', SimpleImputer(strategy='median')),

    ('scale', StandardScaler())
])

# 2. pipeline -> for Categorical columns
categorical_pipeline = Pipeline(steps=[
    ('impute', SimpleImputer(strategy='most_frequent')),
    ('encode', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer(transformers=[
    ('numerical', numeric_pipeline, numerical_cols),
    ('categorical', categorical_pipeline, categorical_cols)
])
```

## Trying Multiple Algorithms

We now compare several models, each wrapped in its own pipeline with the same preprocessor. We evaluate each with 3-fold stratified cross-validation on the training set (never touching the test set yet), so the comparison is fair and robust.

Models tried:

  1. Logistic Regression - simple, interpretablle linear baseline
  2. Decision Tree - a single non-linear model prone to overfitting
  3. Random Forest - an ensemble of trees, usually a strong general-purpose model
  4. Gradient Boosting - sequential ensemble, often most accurate of the classics.

## **SMOTE** (Synthetic Minority Oversampling)

for class imbalance

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.model_selection import cross_val_score
```

```python
# class_weight="Balanced" tell the model to pay more attention to the minority class or rare class (Shared room)
# but the problem is GradientBoosting doesn't support class weight so we left it as it is.


models = {
    "Logistic Regression" : LogisticRegression(class_weight="balanced", random_state=42),
    "Decision Tree"       : DecisionTreeClassifier(class_weight="balanced", random_state=42),
    "Random Forest"       : RandomForestClassifier(class_weight="balanced", random_state=42),
    "Gradient Boosting"   : GradientBoostingClassifier(random_state=42)
}

for name, model in models.items():
  pipe = Pipeline(steps=[("preprocessor", preprocessor), ("classifier", model)])

  # Accuracy alone can be misleading on imbalanced classes, so we use macro f1
  # both: plain accuracy and macro f1 (which treats every class equally)

  accuracy = cross_val_score(pipe, X_train, y_train, cv=3, scoring="accuracy")
  macrof1 = cross_val_score(pipe, X_train, y_train, cv=3, scoring="f1_macro")

  print(f"{name} -> Accuracy: {accuracy.mean():.3f}, F1: {macrof1.mean():.3f}")
```

## Hyperparameter Tuning

- pick best performing model from comparison above (whichever one that turns out to be) and search over its key hyperparameters using `RandomizedSearchCV` (cheaper than an exhaustive grid search, while still exploring the space well). We keep optimizing for macro-F1 since the classes are imabalanced.

```python
from sklearn.model_selection import RandomizedSearchCV

best_pipeline = Pipeline(steps = [
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(class_weight = 'balanced'))
])

param_distribution = {
    "classifier__n_estimators" : [100, 200, 150, 300],
    "classifier__max_depth" : [8, 12, 15, 20, None],
    "classifier__min_samples_split" : [2, 5, 10],
}

search = RandomizedSearchCV(
  estimator = best_pipeline,
  param_distributions = param_distribution,
  n_iter = 10,
  cv = 3,
  scoring = "f1_macro",
  random_state = 42
)

search.fit(X_train, y_train)
```

```python
print("Best parameters: ", search.best_params_)
print("Best CV Macro-F1: ", search.best_score_)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, f1_score

best_pipeline = search.best_estimator_

y_pred = best_pipeline.predict(X_test)

print(f"Accuracy Score: {accuracy_score(y_test, y_pred)}")
print(f"F1 Score: {f1_score(y_test, y_pred, average="macro")}")
```

```python
sns.heatmap(confusion_matrix(y_test, y_pred), annot = True, fmt = ".2f")
```
